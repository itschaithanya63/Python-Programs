```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True
n=int(input("Enter a number"))
is_prime(n)
```

```python
library = []

def add_book(book): library.append(book)
def remove_book(book):
    if book in library: library.remove(book)
    else: print("Book not found!")
def search_book(book): return book in library
def display(): print("Library Books:", library)

while True:
    choice = input("\n1:Add 2:Remove 3:Search 4:Display 5:Exit\nChoice: ")
    if choice == '1': add_book(input("Enter title: "))
    elif choice == '2': remove_book(input("Enter title: "))
    elif choice == '3': print("Found" if search_book(input("Enter title: ")) else "Not Found")
    elif choice == '4': display()
    elif choice == '5': break
```

```python
def analyze_results(marks):
    print(f"Average: {sum(marks)/len(marks)}")
    print(f"Highest: {max(marks)}, Lowest: {min(marks)}")
    print("Passed:", [m for m in marks if m > 40])

marks = [int(input(f"Enter marks for student {i+1}: ")) for i in range(5)]
analyze_results(marks)
```

```python
def is_strong(p):
    if len(p) < 8: return False
    d = any(c.isdigit() for c in p)
    u = any(c.isupper() for c in p)
    l = any(c.islower() for c in p)
    return d and u and l

pwd = input("Enter password: ")
print("Strong" if is_strong(pwd) else "Weak")
```

```python
contacts = {}
def menu():
    while True:
        c = input("\n1:Add 2:Search 3:Delete 4:Display 5:Exit\nChoice: ")
        if c == '1': contacts[input("Name: ")] = input("Phone: ")
        elif c == '2': print(contacts.get(input("Name: "), "Not found"))
        elif c == '3': contacts.pop(input("Name: "), None)
        elif c == '4': print(contacts)
        else: break
menu()
```

```python
balance = 0
def deposit(amt): 
    global balance
    balance += amt
def withdraw(amt): 
    global balance
    if amt <= balance: balance -= amt
    else: print("Insufficient funds")
def check_balance(): print(f"Balance: {balance}")

deposit(500)
withdraw(200)
check_balance()
```

```python
arr = [10, 20, 30, 40, 50]

# Max and Min without built-ins
max_v, min_v = arr[0], arr[0]
for x in arr:
    if x > max_v: max_v = x
    if x < min_v: min_v = x

# Reverse without built-ins
rev = []
for i in range(len(arr)-1, -1, -1):
    rev.append(arr[i])

print(f"Max: {max_v}, Min: {min_v}, Reversed: {rev}")
```
